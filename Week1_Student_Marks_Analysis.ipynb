# Week 1 Project: Analyze Student Marks Dataset
**Topic:** Python for Data Science Basics  
**Objective:** Practice Pandas, basic data cleaning, and descriptive statistics using the Students Performance dataset.

This notebook loads the supplied CSV, checks and cleans missing data, calculates average/minimum/maximum scores, and displays the results.

## 1. Import libraries

```python
import pandas as pd
from pathlib import Path

# Keep StudentsPerformance.csv in the same folder as this notebook.
df = pd.read_csv('StudentsPerformance.csv')
print('Dataset loaded successfully!')
print('Rows and columns:', df.shape)
display(df.head())
```

## 2. Inspect the dataset

```python
print('Column names:')
print(df.columns.tolist())

print('\nDataset information:')
df.info()

print('\nMissing values before cleaning:')
display(df.isnull().sum())
```

## 3. Clean missing data

```python
# Make a copy so the original loaded data remains unchanged.
clean_df = df.copy()

# Fill missing numeric values with the column median and categorical values
# with the most frequent value. If no values are missing, the data remains unchanged.
numeric_cols = clean_df.select_dtypes(include='number').columns
categorical_cols = clean_df.select_dtypes(exclude='number').columns

for col in numeric_cols:
    if clean_df[col].isnull().any():
        clean_df[col] = clean_df[col].fillna(clean_df[col].median())

for col in categorical_cols:
    if clean_df[col].isnull().any():
        mode_values = clean_df[col].mode()
        if not mode_values.empty:
            clean_df[col] = clean_df[col].fillna(mode_values.iloc[0])

print('Missing values after cleaning:')
display(clean_df.isnull().sum())
```

## 4. Calculate average, minimum, and maximum scores

```python
score_cols = ['math score', 'reading score', 'writing score']

score_summary = pd.DataFrame({
    'Average': clean_df[score_cols].mean(),
    'Minimum': clean_df[score_cols].min(),
    'Maximum': clean_df[score_cols].max()
}).round(2)

score_summary.index.name = 'Subject'
display(score_summary)
```

## 5. Display useful results using Pandas

```python
print('Overall score statistics:')
display(clean_df[score_cols].describe().round(2))

print('Average score by gender:')
display(clean_df.groupby('gender')[score_cols].mean().round(2))

print('First 10 cleaned records:')
display(clean_df.head(10))
```

## 6. Save the cleaned dataset and summary

```python
clean_df.to_csv('StudentsPerformance_Cleaned.csv', index=False)
score_summary.to_csv('Student_Score_Summary.csv')
print('Saved: StudentsPerformance_Cleaned.csv')
print('Saved: Student_Score_Summary.csv')
```

## Conclusion
The dataset was loaded and inspected using Pandas. Missing values were checked and handled where present. The average, minimum, and maximum marks were calculated for mathematics, reading, and writing. Group-wise averages and descriptive statistics were also displayed. The supplied dataset contains 1,000 student records and 8 columns; no missing values were found in the original data.
